## 1.5.3 넘파이의 산술 연산
- x,y의 원소 수가 같아야함
- 각 원소에 대해서 계산이 행해짐, (원소별 = element-wise)

```python
x = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 6.0])
print(x + y)
print(x - y)
print(x * y)
print(x / y)
```

## 1.5.4 넘파이의 N차원 배열
- 행렬도 나타낼 수 있음, 행렬의 산술 연산도 대응하는 원소끼리 계산됨


```python
A = np.array([[1, 2], [3, 4]])
print(A)
A.shape
A.dtype
```

```python
B = np.array([[3, 0], [0, 6]])
print(A + B)
print(A * B)
```

- 행렬과 스칼라값의 산술 연산 가능

```python
print(A * 10)
```

## 1.5.5 브로드캐스트
- 형상이 다른 배열끼리도 계산가능
1. 스칼라 값인 10이 2x2 행렬로 확장
2. 1차원 배열 B가 2차원 배열으로 확장

```python
A = np.array([[1, 2], [3, 4]])
print(A*10)
B = np.array([10, 20])
print(A * B) # B가 [[10, 20], [10, 20]]이 됨
```

## 1.5.6 원소 접근


```python
X = np.array([[51, 55], [14, 19], [0, 4]])
print(X)
print(f"0행 : {X[0]}")
print(f"(0,1)의 원소 : {X[0][1]}")

for row in X:
  print(row)
```

```python
X = X.flatten() # 평탄화 - X를 1차원 배열로 반환
print(X)
print(X[np.array([0, 2, 4])])
```

```python
print(X > 15) # 각 원소에 대해서 비교 실행후 bool값 array반환
print(X[X > 15]) # 15이상인 원소만 모으기
```

# 1.6 Matplotlib - 그래프, 시각화
## 1.6.1 단순한 그래프 그리기

```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터 준비
x = np.arange(0, 6, 0.1) # 0부터 6까지 0.1간격으로 생성 [0, 0.1, 0.2, ... , 5.8, 5.9]
y = np.sin(x)

# 그래프 그리기
plt.plot(x, y) # x축 y축 넣어주는 거임
plt.show()
```

## 1.6.2 pyplot의 기능

```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터 준비
x = np.arange(0, 6, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

# 그래프 그리기
plt.plot(x, y1, label="sin")
plt.plot(x, y2, linestyle="--", label="cos") # cos함수는 점선으로 그리기
plt.xlabel("x") # x축 이름
plt.ylabel("y") # y축 이름
plt.title("sin & cos") # 제목
plt.legend()
plt.show()
```
