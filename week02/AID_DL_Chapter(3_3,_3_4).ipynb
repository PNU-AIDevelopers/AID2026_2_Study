시그모이드 함수, 및 항등 함수 정의

---




```python
import numpy as np

# 시그모이드 함수 (3.2절 / 88쪽)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 항등 함수 (3.4.2절 / 89쪽)
def identity_function(x):
    return x
```

1,2 차원 행렬

---

```python
# --- 1차원 배열 ---
A = np.array([1, 2, 3, 4])
print("A = ", A)
print("np.ndim(A) =", np.ndim(A))  # 배열의 차원 수
print("A.shape =", A.shape)        # 배열의 형상 (튜플 반환)
print("A.shape[0] =", A.shape[0])
print("-" * 30)

# --- 2차원 배열 (행렬) ---
B = np.array([[1, 2], [3, 4], [5, 6]])
print("B =\n", B)
print("np.ndim(B) =", np.ndim(B))
print("B.shape =", B.shape)
```

행렬 곱

---

```python
# --- 2x2 행렬의 곱 ---
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A.shape:", A.shape)
print("B.shape:", B.shape)
print("A • B =\n", np.dot(A, B))
print("=" * 30)

# --- 2x3 행렬과 3x2 행렬의 곱 ---
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[1, 2], [3, 4], [5, 6]])

print("A.shape:", A.shape)
print("B.shape:", B.shape)
print("A • B =\n", np.dot(A, B))
print("=" * 30)

# --- 형상이 맞지 않는 경우 (오류 발생 예시) ---
C = np.array([[1, 2], [3, 4]])
print("C.shape:", C.shape)
print("A.shape:", A.shape)


print(np.dot(A, C))
```

간단한 신경망에서의 행렬 곱

---

```python
X = np.array([1, 2])
W = np.array([[1, 3, 5], [2, 4, 6]])

print("X.shape:", X.shape)
print("W.shape:", W.shape)

Y = np.dot(X, W)
print("Y =", Y)
```

3층 신경망 구현

---

```python
# --- 0층 -> 1층 신호 전달 (86쪽~87쪽) ---
X = np.array([1.0, 0.5])
W1 = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
B1 = np.array([0.1, 0.2, 0.3])

print("W1.shape:", W1.shape)
print("X.shape:", X.shape)
print("B1.shape:", B1.shape)

A1 = np.dot(X, W1) + B1
Z1 = sigmoid(A1)

print("A1 =", A1)
print("Z1 (1층 활성화 출력) =", Z1)
print("-" * 40)

# --- 1층 -> 2층 신호 전달 (88쪽) ---
W2 = np.array([[0.1, 0.4], [0.2, 0.5], [0.3, 0.6]])
B2 = np.array([0.1, 0.2])

print("Z1.shape:", Z1.shape)
print("W2.shape:", W2.shape)
print("B2.shape:", B2.shape)

A2 = np.dot(Z1, W2) + B2
Z2 = sigmoid(A2)

print("A2 =", A2)
print("Z2 (2층 활성화 출력) =", Z2)
print("-" * 40)

# --- 2층 -> 3층(출력층) 신호 전달 (89쪽) ---
W3 = np.array([[0.1, 0.3], [0.2, 0.4]])
B3 = np.array([0.1, 0.2])

A3 = np.dot(Z2, W3) + B3
Y = identity_function(A3)  # 혹은 Y = A3

print("A3 =", A3)
print("Y (최종 출력) =", Y)
```

정리

---

```python
def init_network():
    network = {}
    network['W1'] = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
    network['b1'] = np.array([0.1, 0.2, 0.3])
    network['W2'] = np.array([[0.1, 0.4], [0.2, 0.5], [0.3, 0.6]])
    network['b2'] = np.array([0.1, 0.2])
    network['W3'] = np.array([[0.1, 0.3], [0.2, 0.4]])
    network['b3'] = np.array([0.1, 0.2])

    return network

def forward(network, x):
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']

    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)
    a3 = np.dot(z2, W3) + b3
    y = identity_function(a3)

    return y

# 네트워크 초기화 및 추론 실행
network = init_network()
x = np.array([1.0, 0.5])
y = forward(network, x)

print("3층 신경망 최종 출력 Y = ", y)  # [0.31682708, 0.69627909]
```
