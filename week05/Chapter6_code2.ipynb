```python
import numpy as np
import matplotlib.pyplot as plt


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def tanh(x):
    return np.tanh(x)          # 쌍곡탄젠트 (hyperbolic tangent)


def arctan(x):
    return np.arctan(x)        # 아크탄젠트 (역탄젠트), 비교용


# 사용할 활성화 함수를 여기서 선택: sigmoid / tanh / arctan
activation = arctan

x = np.random.randn(1000, 100)  # 1000개의 데이터
node_num = 100                  # 각 은닉층의 노드 수
hidden_layer_size = 5           # 은닉층이 5개
activations = {}                # 활성화값을 저장

for i in range(hidden_layer_size):
    if i != 0:
        x = activations[i-1]

    # w = np.random.randn(node_num, node_num) * 1
    #w = np.random.randn(node_num, node_num) * 0.01
    w = np.random.randn(node_num, node_num) / np.sqrt(node_num)  # Xavier 초깃값

    a = np.dot(x, w)
    z = activation(a)
    activations[i] = z
```

```python
for i, a in activations.items():
    plt.subplot(1, len(activations), i+1)
    plt.title(str(i+1) + "-layer")
    plt.hist(a.flatten(), 30, range=(0, 1))
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


def ReLU(x):
    return np.maximum(0, x)


x = np.random.randn(1000, 100)  # 1000개의 데이터
node_num = 100  # 각 은닉층의 노드 수
hidden_layer_size = 5  # 은닉층이 5개
activations = {}  # 활성화값을 저장

for i in range(hidden_layer_size):
    if i != 0:
        x = activations[i-1]
    #w = np.random.randn(node_num, node_num) / np.sqrt(node_num)
    #w = np.random.randn(node_num, node_num) * 0.01
    w = np.random.randn(node_num, node_num) * np.sqrt(2/node_num)
    a = np.dot(x, w)
    z = ReLU(a)
    activations[i] = z
```

```python
for i, a in activations.items():
    plt.subplot(1, len(activations), i+1)
    plt.title(str(i+1) + "-layer")
    plt.hist(a.flatten(), 30, range=(0, 1))
plt.show()
```
